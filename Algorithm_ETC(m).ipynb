# Algorithm ETC(m) for the Problem of Bandits

The Problem of Bandits is specific Dynamic Programming problem, which can be modelled by a MDP with:
- $\mathcal{S}=\mathcal{A}=\{1,2,\dots,K\}$ machines or arms;
- the rewards are i.i.d. of unknown laws.


The ETC(m) is an algorithm to select the optimal machine, by maximing the rewards.


```python

import numpy as np

# parameters
lambdas = [2, 3, 4]
K = 3
m = 10
T = 1000
N = 100

rng = np.random.default_rng(2026)           # random generator with seed 2026

optimal_machines = []                       # accounts for the optimal machines among 1, 2, ..., K
total_rewards = []                          # accounts for the total rewards


for essai in range(N):                      # 100 essai
                                            # find best machine + total reward, for each essai

    # Phase 1 : exploration
    recompenses = []

    for i in range(K):
        observations = rng.poisson(lambdas[i], m)      # m poisson experiments of parameter lambda_i / m rewards from machine i
        recompenses.append(observations)

    # Average rewards from exploration: \hat{\mu}
    moyennes = np.array([ np.mean(recompenses[i]) for i in range(K) ])


    # Phase 2 : exploitation
    best_machine = np.argmax(moyennes)                 # best machine from explorative phase

    # Total reward obtained from exploration
    total = sum(np.sum(x) for x in recompenses)

    # Total reward after exploration and exploitation, until the horizon T
    total += np.sum(
        rng.poisson(
            lambdas[best_machine],
            T - K*m
        )
    )

    optimal_machines.append(best_machine)       # save best machine in 0, 1, ..., K-1
    total_rewards.append(total)                 # save total reward


# Statistics
optimal_machines = np.array(optimal_machines)
total_rewards = np.array(total_rewards)

for i in range(K):
    print(f"Machine {i+1} : the best in {np.sum(optimal_machines == i)}% of times")

print("Average total reward: ", np.mean(total_rewards))

print("Average reward for each trial :", np.mean(total_rewards) / T)

print("Standard deviation :", np.std(total_rewards, ddof=1))

```
